```python
from src.evaluation.evaluation_l1_util import Span, extract_spans, _get_unmatched_system, CONTEXT_CHARS
from typing import List, Tuple, Optional
def _match_and_get_partner(query_spans: List[Span], candidate_spans: List[Span]) -> List[Tuple[Span, Optional[Span]]]:
    """
    Same greedy matching rule as match_spans()/Span.matches(), but returns the
    actual matched partner (or None) for each query span, instead of just a count.
    """
    used = [False] * len(candidate_spans)
    results = []
    for q in query_spans:
        matched = None
        for j, c in enumerate(candidate_spans):
            if not used[j] and q.matches(c):
                used[j] = True
                matched = c
                break
        results.append((q, matched))
    return results


def export_unmatched_for_review(
    triple_h_hllm_llm: List[Tuple[str, str, str]],
    context_chars: int = CONTEXT_CHARS,
    output_csv: Optional[str] = None,
) -> List[dict]:
    """
    For every Human+LLM span that does NOT match the independent Human gold,
    export a row for manual adjudication: doc, label, ids, text, whether it's
    identical to raw LLM output, and the candidate Human alternative(s) it
    disagrees with (if any) — so a reviewer can judge who was right.
    """
    rows = []

    for human_path, hllm_path, llm_path in triple_h_hllm_llm:
        doc_name = Path(human_path).stem
        human_spans = extract_spans(human_path, context_chars)
        hllm_spans  = extract_spans(hllm_path,  context_chars)
        llm_spans   = extract_spans(llm_path,   context_chars)

        all_labels = sorted({s.labelname for s in human_spans} | {s.labelname for s in hllm_spans})

        for label in all_labels:
            g    = [s for s in human_spans if s.labelname == label]
            sys_ = [s for s in hllm_spans  if s.labelname == label]
            llm_ = [s for s in llm_spans   if s.labelname == label]

            unmatched_hllm  = _get_unmatched_system(g, sys_)   # Human+LLM spans Human gold doesn't confirm
            unmatched_human = _get_unmatched_system(sys_, g)   # Human spans Human+LLM doesn't confirm (FN)

            partners = _match_and_get_partner(unmatched_hllm, llm_)

            alt_text = " | ".join(
                f"[id={h.attributes.get('id','?')}] {h.text[:80]}" for h in unmatched_human
            )

            for hllm_span, llm_match in partners:
                rows.append({
                    "doc":            doc_name,
                    "label":          label,
                    "hllm_id":        hllm_span.attributes.get("id", ""),
                    "hllm_text":      hllm_span.text,
                    "hllm_context":   hllm_span.context_text[:50],
                    "matches_llm":    llm_match is not None,
                    "llm_id":         llm_match.attributes.get("id", "") if llm_match else "",
                    "human_alt_spans": alt_text,   # candidate Human FN spans for this doc/label, for comparison
                    "verdict":        "",          # fill manually: human / human_llm / both / neither
                })

    if output_csv and rows:
        import csv
        with open(output_csv, "w", newline="", encoding="utf-8") as f:
            writer = csv.DictWriter(f, fieldnames=list(rows[0].keys()))
            writer.writeheader()
            writer.writerows(rows)
        print(f"Wrote {len(rows)} rows to {output_csv}")

    return rows

csv = export_unmatched_for_review(triple_h_hllm_llm)
```

```python
import numpy as np


def bootstrap_micro_f1(results, n_bootstrap=10000, seed=42):
    per_doc = results["per_document"]

    documents = list(per_doc.keys())
    n_docs = len(documents)

    rng = np.random.default_rng(seed)

    bootstrap_f1s = np.empty(n_bootstrap)

    for i in range(n_bootstrap):
        # Sample documents WITH replacement
        sampled_docs = rng.choice(
            documents,
            size=n_docs,
            replace=True
        )

        # Sum raw counts
        tp = sum(per_doc[doc]["tp"] for doc in sampled_docs)
        n_gold = sum(per_doc[doc]["n_gold"] for doc in sampled_docs)
        n_system = sum(per_doc[doc]["n_system"] for doc in sampled_docs)

        # Micro F1
        precision = tp / n_system if n_system > 0 else 0
        recall = tp / n_gold if n_gold > 0 else 0

        if precision + recall > 0:
            f1 = 2 * precision * recall / (precision + recall)
        else:
            f1 = 0

        bootstrap_f1s[i] = f1

    return bootstrap_f1s
```

```python
bootstrap_f1s = bootstrap_micro_f1(
    results_h_vs_llm,
    n_bootstrap=10000,
    seed=42
)
observed_f1 = results_h_vs_llm["overall"]["F1"]
ci_low, ci_high = np.percentile(
    bootstrap_f1s,
    [2.5, 97.5]
)

print(f"Micro F1: {observed_f1:.3f}")
print(f"95% CI: [{ci_low:.3f}, {ci_high:.3f}]")
```
